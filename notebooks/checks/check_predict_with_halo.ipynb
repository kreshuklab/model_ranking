```python
from typing import Dict, Any
import numpy as np

from torch_em.util.prediction import predict_with_halo
from model_ranking import load_h5

from pytorch3dunet.unet3d.model import get_model
from pytorch3dunet.unet3d.utils import load_checkpoint
```

```python
VNC_path = "/scratch/talks/data/VNC/resized_pixels/test.h5"
VNC_data = load_h5(VNC_path, "raw")
```

```python
print(VNC_data.shape)
```

```python
model_cfg: Dict[str, Any] =  {
    "name": "UNet2d_as3d",
    "in_channels": 1,
    "out_channels": 1,
    "layer_order": "bcr",
    "f_maps": 32,
    "final_sigmoid": True,
    "is_segmentation": True
}
VNC_model_path = "/g/kreshuk/talks/segmentation_ModelSelection/experiments/VNC/BatchNorm/V_model1/best_checkpoint.pytorch"
VNC_model = get_model(model_cfg)
load_checkpoint(VNC_model_path, VNC_model)
```

```python
pred= predict_with_halo(
    VNC_data,
    VNC_model,
    gpu_ids=['cpu'],
    block_shape=(1,256,256),
    halo=(0,32,32),
    output=None,
    blockwise_prediction=True
)
```
