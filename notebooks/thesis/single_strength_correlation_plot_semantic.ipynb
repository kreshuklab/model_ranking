## Comparison: Original vs Shape-Based

**Original function** (`plot_single_consistency_vs_performance_CVPR`):
- Each model gets a different shade of color within its source family
- Example: EPFL models get progressively darker blues
- Good for: Distinguishing individual models within a source

**Shape-based function** (`plot_single_consistency_vs_performance_CVPR_shapes`):
- All models from same source get the exact same color
- Model architecture is indicated by shape (circle, square, diamond, triangle)
- Good for: Comparing architectures across sources, cleaner color grouping

**Legend:**
- Circle (○): U-Net (No Augs) - contains "NA"
- Square (■): ResUNet - contains "Res"
