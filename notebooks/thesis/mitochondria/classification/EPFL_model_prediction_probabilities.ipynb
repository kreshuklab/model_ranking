```python
labels = load_h5(EtoR_ResNet50_pred_path, "labels")
```

```python
th = 0.5
EtoR_MN3_hard_preds = (EtoR_MN3_preds > th).astype(np.int8)
EtoR_MN2_hard_preds = (EtoR_MN2_preds > th).astype(np.int8)
HtoR_MN2_hard_preds = (HtoR_MN2_preds > th).astype(np.int8)
RtoR_MN2_hard_preds = (RtoR_MN2_preds > th).astype(np.int8)
```

```python
# False negative ratios
EtoR_MN3_fnr = np.sum((EtoR_MN3_hard_preds == 0) & (labels == 1)) / np.sum(labels == 1)
EtoR_MN2_fnr = np.sum((EtoR_MN2_hard_preds == 0) & (labels == 1)) / np.sum(labels == 1)
HtoR_MN2_fnr = np.sum((HtoR_MN2_hard_preds == 0) & (labels == 1)) / np.sum(labels == 1)
RtoR_MN2_fnr = np.sum((RtoR_MN2_hard_preds == 0) & (labels == 1)) / np.sum(labels == 1)
print(f"False negative ratios: \n EtoR_MN3: {EtoR_MN3_fnr:.3f}, \n EtoR_MN2: {EtoR_MN2_fnr:.3f}, \n HtoR_MN2: {HtoR_MN2_fnr:.3f}, \n RtoR_MN2: {RtoR_MN2_fnr:.3f}")
```

```python
# False positive ratios
EtoR_MN2_fpr = np.sum((EtoR_MN2_hard_preds == 1) & (labels == 0)) / np.sum(labels == 0)
HtoR_MN2_fpr = np.sum((HtoR_MN2_hard_preds == 1) & (labels == 0)) / np.sum(labels == 0)
RtoR_MN2_fpr = np.sum((RtoR_MN2_hard_preds == 1) & (labels == 0)) / np.sum(labels == 0)
print(f"False positive ratios: EtoR_MN2: {EtoR_MN2_fpr:.3f}, HtoR_MN2: {HtoR_MN2_fpr:.3f}, RtoR_MN2: {RtoR_MN2_fpr:.3f}")
```
