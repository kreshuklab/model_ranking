```python
base_path = "/g/kreshuk/talks/sampled_features/classification/mitochondria/transfer_metric_results/fullset_results2"
result_paths = Path(base_path).glob("*.json")


transfer_metric_names_mito = []
transfer_metrics_avg_dfs_mito = []
transfer_metrics_dfs_mito = []
for result_path in result_paths:
    results = load_transfer_metric_results(result_path)
    # correlation_scores = results["correlation_scores"]
    # idx = 0
    # per_target_KT = np.array(correlation_scores["KT"])
    # per_target_SP = np.array(correlation_scores["SP"])
    # per_target_PE = np.array(correlation_scores["PE"])
    transfer_scores = results["transfer_scores"]
    performance_scores_mito = results["performance_scores"]
    targets = list(transfer_scores.keys())
    KT_scores, SP_scores, PE_scores = to_target_transfer_correlations(
        targets=targets,
        transfer_metric_per_target=transfer_scores,
        performance_per_target=performance_scores_mito,
        invert_transfer_metric=False,
        )
    # df = correlation_table(per_target_KT, per_target_SP, per_target_PE, targets = ["EPFL", "Hmito", "Rmito"])
    df = correlation_table(KT_scores, SP_scores, PE_scores, targets=targets)
    transfer_metric_abbrev = transfer_metric_abbrevs[result_path.stem]
    transfer_metric_names_mito.append(transfer_metric_abbrev)
    df_avg = avg_correlation_table(df)
    transfer_metrics_avg_dfs_mito.append(df_avg)
    transfer_metrics_dfs_mito.append(df)
```

```python
avg_dataframes = [
    df_avg_mito_gauss_EI,
    df_avg_mito_gauss_NHD,
    transfer_metrics_avg_dfs_mito
]
```

```python
import numpy as np
import pandas as pd

dataframes = [
    df_mito_gauss_EI,
    df_mito_gauss_NHD,
    transfer_metrics_dfs_mito[7:],
]

avg_dataframes = [
    df_avg_mito_gauss_EI,
    df_avg_mito_gauss_NHD,
    transfer_metrics_avg_dfs_mito[7:],
]

metric_names = [
    "Gauss_EI",
    "Gauss_NHD",
] + transfer_metric_names_mito[7:]

# Flatten the nested lists of dataframes so each metric lines up with one score df and one avg df
flat_dataframes = dataframes[:-1] + list(dataframes[-1])
flat_avg_dataframes = avg_dataframes[:-1] + list(avg_dataframes[-1])

if len(flat_dataframes) != len(metric_names):
    raise ValueError(
        f"Length mismatch: {len(flat_dataframes)} dataframes vs {len(metric_names)} metric names"
    )

if len(flat_avg_dataframes) != len(metric_names):
    raise ValueError(
        f"Length mismatch: {len(flat_avg_dataframes)} avg dataframes vs {len(metric_names)} metric names"
    )

target_columns = ["EPFL", "Hmito", "Rmito", "VNC"]
corr_specs = [
    ("kt", "kt pval", "KT_avg", "KT_std"),
    ("s rho", "s rho pval", "SR_avg", "SR_std"),
    ("pr", "pr pval", "PR_avg", "PR_std"),
]

combined_rows = []

for metric_name, df_metric, df_metric_avg in zip(metric_names, flat_dataframes, flat_avg_dataframes):
    df_reset = df_metric.reset_index()

    for corr_type, pval_col, avg_col, std_col in corr_specs:
        corr_row = {
            "metric_name": metric_name,
            "corr_type": corr_type,
            "pval_type": "score",
            "corr_avg": df_metric_avg[avg_col].iloc[0],
            "corr_std": df_metric_avg[std_col].iloc[0],
        }
        pval_row = {
            "metric_name": metric_name,
            "corr_type": corr_type,
            "pval_type": "pval",
            "corr_avg": np.nan,
            "corr_std": np.nan,
        }

        for target in target_columns:
            target_match = df_reset[df_reset["targets"] == target]
            if target_match.empty:
                corr_row[target] = np.nan
                pval_row[target] = np.nan
            else:
                corr_row[target] = target_match[corr_type].iloc[0]
                pval_row[target] = target_match[pval_col].iloc[0]

        combined_rows.append(corr_row)
        combined_rows.append(pval_row)

combined_results_df = pd.DataFrame(combined_rows)

sort_orders = {
    "kt": 0,
    "s rho": 1,
    "pr": 2,
    "score": 0,
    "pval": 1,
}

combined_results_df = combined_results_df.sort_values(
    by=["metric_name", "corr_type", "pval_type"],
    key=lambda s: s.map(sort_orders).fillna(s),
).reset_index(drop=True)

combined_results_df
```
