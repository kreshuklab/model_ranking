# Style Examples

## Available matplotlib styles:
- `"default"` - Standard matplotlib style
- `"seaborn-v0_8-colorblind"` - Colorblind-friendly palette
- `"seaborn-v0_8-whitegrid"` - White background with grid
- `"seaborn-v0_8-darkgrid"` - Dark grid background
- `"ggplot"` - ggplot2-inspired style
- `"bmh"` - Bayesian Methods for Hackers style
- `"fivethirtyeight"` - FiveThirtyEight style

## Legend placement options:
- **Outside plot**: Leave `legend_loc=None` (default behavior)
- **Inside plot**: Use `legend_loc="upper left"`, `"upper right"`, `"lower left"`, `"lower right"`, `"center"`, `"best"`, etc.
- **Custom position**: Use `legend_bbox_to_anchor=(x, y)` for precise control

## Correlation box placement:
- **Lower left**: `corr_box_loc=(0.05, 0.05)`
- **Lower right**: `corr_box_loc=(0.72, 0.05)` or `(0.95, 0.05)` with right alignment
- **Upper left**: `corr_box_loc=(0.05, 0.95)`
- **Upper right**: `corr_box_loc=(0.72, 0.95)`

# Quick Reference Guide

## Summary of Modifications

### ✅ What's New:

1. **Greek Symbols in Correlation Box**
   - Kτ (Kendall tau), Sρ (Spearman rho), Pr (Pearson r)
   - Dotted border instead of solid
   - Values indented for readability

2. **Flexible Legend Placement**
   - `legend_loc=None` → Outside plot (default, good for large legends)
   - `legend_loc="upper left"` → Inside plot (good for small legends)
   - See Example 1 vs Example 3 for comparison

3. **Customizable Correlation Box Position**
   - `corr_box_loc=(x, y)` where x, y are in [0, 1]
   - Lower left: `(0.05, 0.05)`
   - Lower right: `(0.72, 0.05)`
   - Upper corners: adjust y to `0.95`

4. **Tick Label Font Control**
   - `tick_fontsize=14` for independent tick label sizing
   - If not set, defaults to `fontsize - 4`

5. **Plot Style Selection**
   - `plot_style="seaborn-v0_8-colorblind"` for publications
   - `plot_style="ggplot"` for presentations
   - Many other styles available!

### 🎯 Common Use Cases:

| Scenario | Settings |
|----------|----------|
| Large legend | `legend_loc=None` (default) |
| Small legend | `legend_loc="upper left"` or `"best"` |
| Colorblind-friendly | `plot_style="seaborn-v0_8-colorblind"` |
| Clean grid | `plot_style="seaborn-v0_8-whitegrid"` |
| Avoid data overlap | Adjust `corr_box_loc` to empty corner |
