```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
```

```python
# Define 3D Gaussian function
def gaussian_3d(x, y, mu_x=0, mu_y=0, sigma_x=1, sigma_y=1):
    """
    Calculate 3D Gaussian (bell curve) values.
    
    Parameters:
    - x, y: coordinate arrays
    - mu_x, mu_y: mean values for x and y
    - sigma_x, sigma_y: standard deviations for x and y
    """
    return (1 / (2 * np.pi * sigma_x * sigma_y)) * \
           np.exp(-((x - mu_x)**2 / (2 * sigma_x**2) + 
                    (y - mu_y)**2 / (2 * sigma_y**2)))

# Create grid
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Calculate Gaussian values
Z = gaussian_3d(X, Y, mu_x=0, mu_y=0, sigma_x=1.7, sigma_y=1.7)
```

```python
# Create 3D plot
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection='3d')

# Plot surface
surf = ax.plot_surface(X, Y, Z, cmap=cm.plasma, 
                       linewidth=0, antialiased=True, alpha=0.8)

# Customize plot
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_zlabel('Probability Density', fontsize=12)
ax.set_title('3D Gaussian Distribution', fontsize=14, pad=20)

# Add colorbar
fig.colorbar(surf, shrink=0.5, aspect=5)

# Adjust viewing angle
ax.view_init(elev=30, azim=45)

plt.tight_layout()

# Save as SVG
output_path = '/g/kreshuk/talks/model_ranking/notebooks/CVPR_Figures/gaussian_3d/gaussian_3d_2.svg'
plt.tight_layout()
plt.savefig(output_path, format='svg', bbox_inches='tight', dpi=300)
print(f"✓ Figure saved to: {output_path}")
plt.show()
```
