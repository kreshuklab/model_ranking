```python
import numpy as np
from tqdm import tqdm
from pathlib import Path
from model_ranking import (
    load_h5
)
from pytorch3dunet.unet3d.predictor import pmaps_to_IN_seg
```

```python
path = "/g/kreshuk/talks/consistency_results/Instance_segmentation/Cells/FlyWing_to_Ovules_gap/consistency/P_full/fw_model_Res2/norm_50_950/gauss_a01-02/predictions/N_294_final_crop_ds2_predictions.h5"
preds = load_h5(path, "predictions")
segs = np.zeros(preds.squeeze().shape, dtype=np.uint16)
for i, pred in enumerate(tqdm(preds)):
    seg = pmaps_to_IN_seg(
        pred.squeeze(),
        min_size=50,
        zero_largest_instance=False,
        zero_large_instances=True,
        large_instance_multiplier=1.7,
        beta=0.5,
        max_obj_size=5867
    )
    segs[i] = seg
```
