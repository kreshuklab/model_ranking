```python
import json
import matplotlib.pyplot as plt
import numpy as np
```

```python
path = "/g/kreshuk/talks/toothfairy/preds/noisy/sigma_1.5x/Dataset666_ToothFairy2/inferTs_nnUNetTrainerSwinUMambaD__nnUNetPlans__2d__0/AdaRandError_consistency.json"
with open(path, 'r') as f:
    data = json.load(f)
```

```python
AdaRand_consis_scores = []
for id, scores in data.items():
    are, prec, rec = scores
    AdaRand_consis_scores.append(1 - are)
```

```python
_ = plt.hist(AdaRand_consis_scores, bins=20)
_ = plt.xlabel("Consistency score (1 - AdaRandError)")
_ = plt.ylabel("Number of images")
_ = plt.title("Consistency scores SwinUMambaD (sigma=1.5x) ToothFairy2 test images")
_ = plt.grid()
_ = plt.show()
```
