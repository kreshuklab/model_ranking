```python
from model_ranking import (
    load_h5,
    calculate_EI_binary,
)
import numpy as np
```

```python
p_path = "/g/kreshuk/talks/sampled_features/classification/mitochondria/EPFL_to_EPFL/DenseNet121_EPFL/Gauss/a02-025/predictions.h5"
unp_path = "/g/kreshuk/talks/sampled_features/classification/mitochondria/EPFL_to_EPFL/DenseNet121_EPFL/None/predictions.h5"

preds_p = load_h5(p_path, "predictions")
preds_unp = load_h5(unp_path, "predictions")
```

```python
a = np.array([0, 1, 1])
b = np.array([1, 0, 1])
a_probs = np.array([0.1, 0.6, 0.8])
b_probs = np.array([0.9, 0.4, 0.8])
c = np.stack((a, b))
c_probs = np.stack((a_probs, b_probs))
print(c.shape)
print(c_probs.shape)
```

```python
EI, EI_mean, _, _ = calculate_EI_binary(c, c_probs)
print(EI)
print(EI.shape)
print(EI_mean)
```

```python
preds = np.stack((preds_p>0.5, preds_unp>0.5))
preds_probs = np.stack((preds_p, preds_unp))
EI, EI_mean, _, _ = calculate_EI_binary(preds, preds_probs)
print(EI.shape)
print(EI_mean)
```
