```python
from pathlib import Path

from model_ranking import (
    plot_performance_vs_transfer_metric_multi_target,
    to_target_transfer_correlations,
    correlation_table,
    load_transfer_metric_results,
)
```

```python
#base_path = Path("/scratch/talks/sampled_features/semantic_segmentation/mitochondria/1k_pixels_sampled/transfer_metric_results")
base_path = Path("/g/kreshuk/talks/sampled_features/semantic_segmentation/nuclei/1k_pixels_sampled/transfer_metric_results/direct_transfer_results")
transfer_metrics = ["GBC", "LEEP", "Gaussian_LEEP", "Hscore", "Regularized_Hscore", "NCTI", "LogME"]

for transfer_metric in transfer_metrics:
    path = base_path / f"nuclei_{transfer_metric}.json"
    results = load_transfer_metric_results(path)
    targets = results['metadata']['targets']
    transfer_scores = results['transfer_scores']
    performance_scores = results['performance_scores']
    plot_performance_vs_transfer_metric_multi_target(
        transfer_scores=transfer_scores,
        performance_scores=performance_scores,
        transfer_metric=transfer_metric
    )
    KT_scores, SP_scores, PE_scores = to_target_transfer_correlations(
        targets=targets,
        transfer_metric_per_target=transfer_scores,
        performance_per_target=performance_scores,
    )
    correlation_df = correlation_table(KT_scores, SP_scores, PE_scores, targets=targets)
    print(correlation_df)
```
